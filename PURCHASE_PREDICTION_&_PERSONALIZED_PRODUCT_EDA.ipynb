```python
total_users = df["user_id"].nunique()

purchasing_users = (
    df.loc[
        df["event_type"] == "purchase",
        "user_id"
    ].nunique()
)

non_purchasing_users = (
    total_users - purchasing_users
)

print("Total Users:", total_users)
print("Purchasing Users:", purchasing_users)
print("Non-Purchasing Users:", non_purchasing_users)
```

```python
purchase_rate = (
    purchasing_users / total_users * 100
)

print(
    "Percentage of users who made a purchase:",
    round(purchase_rate, 2),
    "%"
)
```

```python
customer_features = pd.crosstab(
    df["user_id"],
    df["event_type"]
).reset_index()

for col in ["view", "cart", "purchase"]:
    if col not in customer_features.columns:
        customer_features[col] = 0

customer_features.head()
```

```python
def identify_stage(row):

    if row["purchase"] > 0:
        return "Purchased"

    elif row["cart"] > 0:
        return "Considering"

    else:
        return "Exploring"


customer_features["journey_stage"] = (
    customer_features.apply(
        identify_stage,
        axis=1
    )
)

customer_features.head()
```

```python
journey_counts = (
    customer_features["journey_stage"]
    .value_counts()
)

print(journey_counts)
```

```python
plt.figure(figsize=(8, 5))

journey_counts.plot(kind="bar")

plt.title("Customer Purchase Journey Stage")
plt.xlabel("Journey Stage")
plt.ylabel("Number of Customers")

plt.xticks(rotation=0)

plt.show()
```
